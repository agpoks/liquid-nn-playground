# RNN -- vanilla (Elman) baseline

Elman, *Finding Structure in Time*, Cognitive Science, 1990.

`h_t = tanh(W_x x_t + W_h h_{t-1} + b)` -- no notion of elapsed time at all: no per-unit time constant, no gate, no ODE. This is the baseline every other model in this repo (CT-RNN, Neural ODE, and the liquid networks) adds machinery on top of. See `model.py` and `../../papers/README.md` for the reference.

This notebook trains an `RNNModel` on the UCI Ozone Level Detection dataset, the same task `models/ltc/example.ipynb` uses, so the two are directly comparable.

```python
import sys
sys.path.insert(0, '../..')
sys.path.insert(0, '.')

import torch
import torch.nn as nn
import matplotlib.pyplot as plt

from liquid_playground.data import load_ozone
from liquid_playground.device import resolve_device
from liquid_playground.utils.seed import set_seed
from model import RNNModel

set_seed(0)
# device options: 'auto' (default, picks cuda/mps if available), 'cpu', 'cuda', 'mps'
device = resolve_device('auto')
print('device:', device)
```

```python
train_x, train_y, test_x, test_y = load_ozone()
train_x = train_x.unsqueeze(-1).to(device)
test_x = test_x.unsqueeze(-1).to(device)
train_y, test_y = train_y.to(device), test_y.to(device)
print(train_x.shape, train_y.shape)
```

```python
model = RNNModel(input_size=1, hidden_size=32, output_size=1).to(device)
opt = torch.optim.Adam(model.parameters(), lr=1e-2)
loss_fn = nn.BCEWithLogitsLoss()

history = {'train_loss': [], 'test_acc': []}
epochs = 40
for epoch in range(epochs):
    model.train()
    opt.zero_grad()
    logits = model(train_x).squeeze(-1)
    loss = loss_fn(logits, train_y)
    loss.backward()
    opt.step()

    model.eval()
    with torch.no_grad():
        test_acc = ((model(test_x).squeeze(-1) > 0).float() == test_y).float().mean().item()
    history['train_loss'].append(loss.item())
    history['test_acc'].append(test_acc)

print(f"final test accuracy: {history['test_acc'][-1]:.3f}")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].plot(history['train_loss']); axes[0].set_title('train loss'); axes[0].set_xlabel('epoch')
axes[1].plot(history['test_acc']); axes[1].set_title('test accuracy'); axes[1].set_xlabel('epoch')
fig.tight_layout()
plt.show()
```
